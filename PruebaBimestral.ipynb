```python
import pandas as pd
productos = pd.read_csv('productos_.csv')
clientes = pd.read_csv('clientes.csv')
ventas = pd.read_csv('ventas.csv')
```

1.- Líneas, tickets y unidades. ¿Cuántas líneas de venta, cuántos tickets distintos y cuántas unidades vendidas hay en el archivo? Explique qué representa cada conteo.

```python
print("Líneas de venta:", len(ventas))
print("Tickets distintos:", ventas['id_ticket'].nunique())
print("Unidades vendidas:", ventas['cantidad'].sum())
```

2.-Tickets por sucursal. ¿Cuántos tickets se registraron en cada sucursal? ¿Cuál registró más?

```python
ticketsSuc = ventas.groupby('sucursal')['id_ticket'].nunique()
print(ticketsSuc)
print("Sucursal con más tickets:", ticketsSuc.idxmax())

```

3.-Métodos de pago. ¿Cuántos tickets se pagaron con cada método y cuántos tienen ese dato sin registrar?

```python
tickets = ventas[['id_ticket', 'metodo_pago']].drop_duplicates()
tickets['metodo_pago'] = tickets['metodo_pago'].fillna('Sin registrar')
tabla_MetodosPago = tickets['metodo_pago'].value_counts()
print(tabla_MetodosPago)
```

4.- Precios de lista por categoría. ¿Cuál es el precio de lista mínimo, máximo y promedio de los productos de cada categoría? Use el catálogo de productos.

```python
grupos = productos.groupby('categoria')['precio_lista']
resumen = grupos.agg(['min', 'max', 'mean'])
print (resumen)
```

5.- Unidades por sucursal y categoría. ¿Cuántas unidades vendió cada sucursal de cada categoría? Presente una tabla cruzada con totales por fila y por columna. Compruebe que el total general coincide con la suma de cantidad en ventas.csv.

```python
ventas_prod = pd.merge(ventas, productos)
resumen = ventas_prod.groupby(["sucursal", "categoria"])["cantidad"].sum()
print(resumen)
total_general = ventas["cantidad"].sum()
print("Total en ventas.csv:", total_general)

```

6.-Importe de ventas por sucursal. ¿Cuál fue el importe total de ventas de cada sucursal? Calcule primero el precio total de cada línea como cantidad × precio_unitario.

```python
ventas['total_linea'] = ventas.cantidad * ventas.precio_unitario
importe_suc = ventas.groupby('sucursal')['total_linea'].sum()
print(importe_suc)
```

7.- Tamaño de las compras. ¿Cuántos tickets corresponden a compras de hasta 7 unidades, de 8 a 14 unidades y de 15 o más unidades? Sume primero las cantidades de todas las líneas de cada ticket.

```python
unidades_ticket = ventas.groupby('id_ticket')['cantidad'].sum()

HastaSiete = (unidades_ticket <= 7).sum()
EntreOchoYCatorce = ((unidades_ticket >= 8) & (unidades_ticket <= 14)).sum()
DeQuinceOMas = (unidades_ticket >= 15).sum()

print("Hasta 7 unidades:", HastaSiete)
print("De 8 a 14 unidades:", EntreOchoYCatorce)
print("15 o más unidades:", DeQuinceOMas)
```

8.-Ventas por mes. ¿Cuántos tickets y qué importe total se registraron en cada mes? ¿En qué mes se alcanzó el mayor importe?

```python
ventas['mes'] = ventas['fecha'].astype(str).str[:7]

tickets = ventas.groupby('mes')['id_ticket'].nunique()
importe = ventas.groupby('mes')['total_linea'].sum()

print("Tickets por mes:")
print(tickets)

print("Importe por mes:")
print(importe)

print("Mes con mayor importe:", importe.sort_values().index[-1])
```

10.- Compras por tipo de cliente. Para cada tipo de cliente, ¿cuántos clientes distintos compraron, cuántos tickets generaron y cuál fue el importe total de sus compras?

```python
ventas_cli = pd.merge(ventas, clientes, left_on='id_cliente', right_on='id')

clientes_distintos = ventas_cli.groupby('tipo_cliente')['id_cliente'].nunique()
tickets = ventas_cli.groupby('tipo_cliente')['id_ticket'].nunique()
total_lineas = ventas_cli.groupby('tipo_cliente')['total_linea'].sum()

print("Clientes distintos:", clientes_distintos)
print("Tickets generados:", tickets)
print("Total generado: ", total_lineas)

```
